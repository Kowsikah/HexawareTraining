```python
patient_appointments_df = patient_df.join(appointments_df, patient_df.patient_id == appointments_df.patient_id, "inner")
total_consult_per_patient_city = patient_appointments_df.groupBy("city", patient_df.patient_id, "patient_name").agg(sum("consult").alias("total_consult"))
window_spec_city = Window.partitionBy("city").orderBy(col("total_consult").desc())
highest_spending_patient_per_city = total_consult_per_patient_city.withColumn("rank", rank().over(window_spec_city))
highest_spending_patient_per_city.filter(col("rank") == 1).show()
```

```python
patient_appointments_df = patient_df.join(appointments_df, patient_df.patient_id == appointments_df.patient_id, "inner")
total_consult_per_patient_city = patient_appointments_df.groupBy("city", patient_df.patient_id, "patient_name").agg(sum("consult").alias("total_consult"))
window_spec_city = Window.partitionBy("city").orderBy(col("total_consult"))
lowest_spending_patient_per_city = total_consult_per_patient_city.withColumn("rank", rank().over(window_spec_city))
lowest_spending_patient_per_city.filter(col("rank") == 1).show()
```

```python
window_spec_running_total = Window.partitionBy("patient_id").orderBy("appointment_id").rowsBetween(Window.unboundedPreceding, Window.currentRow)
running_total_fees = appointments_df.withColumn("running_total_fees", sum("consult").over(window_spec_running_total))
running_total_fees.show()
```

```python
window_spec_patient_order = Window.partitionBy("patient_id").orderBy("appointment_id")
appointments_with_lead = appointments_df.withColumn("next_consult_fee", lead(col("consult"), 1).over(window_spec_patient_order))
print("Appointments with Next Consult Fee (using lead()):")
appointments_with_lead.select("patient_id", "appointment_id", "consult", "next_consult_fee").show()
```

```python
appointments_with_lag = appointments_df.withColumn("previous_consult_fee", lag(col("consult"), 1).over(window_spec_patient_order))
print("\nAppointments with Previous Consult Fee (using lag()):")
appointments_with_lag.select("patient_id", "appointment_id", "consult", "previous_consult_fee").show()
```

```python
from pyspark.sql.functions import count,mean,col,sum
from pyspark.sql.types import IntegerType
appointments_by_department = appointments_df.groupBy("department").agg(count("appointment_id").alias("total_appointments"))
appointments_by_department.show()
avg_consult_by_department = appointments_df.groupBy("department").agg(mean("consult").alias("average_consultation_fee"))
avg_consult_by_department.show()
patient_count_by_city = patient_df.groupBy("city").agg(count("patient_id").alias("patient_count"))
patient_count_by_city.show()
appointment_status_distribution = appointments_df.groupBy("status").agg(count("appointment_id").alias("appointment_count"))
appointment_status_distribution.show()
avg_age_by_city = patient_df.groupBy("city").agg(mean("age").alias("average_age"))
avg_age_by_city.show()
top_doctors = appointments_df.groupBy("doctor_name").agg(count("appointment_id").alias("total_appointments"))
top_doctors.orderBy(col("total_appointments").desc()).limit(3).show()
insurance_consult_fees = patient_df.join(appointments_df, on="patient_id", how="inner") \
                                   .withColumn("consult_numeric", col("consult").cast(IntegerType())) \
                                   .groupBy("insurance_status") \
                                   .agg(sum("consult_numeric").alias("total_consultation_fees"))
insurance_consult_fees.show()
```
