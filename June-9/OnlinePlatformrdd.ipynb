```python
total_activities = rdd.count()
freq_rdd = rdd.map(lambda x: (x, 1)).reduceByKey(lambda a, b: a + b).cache()
freq_rdd.collect()
```

```python
most_common = freq_rdd.max(key=lambda x: x[1])
print(f"Most common activity: {most_common[0]} ({most_common[1]} times)")
```

```python
least_common = freq_rdd.min(key=lambda x: x[1])
print(f"Least common activity: {least_common[0]} ({least_common[1]} times)")
```

```python
login_count = rdd.filter(lambda x: x == "LOGIN").count()
login_percentage = (login_count / total_activities) * 100
print(f"LOGIN Percentage: {login_percentage:.2f}%")
```

```python
login_count = rdd.filter(lambda x: x == "VIDEO_PLAY").count()
login_percentage = (login_count / total_activities) * 100
print(f"VIDEO_PLAY Percentage: {login_percentage:.2f}%")
```

```python
report = rdd.collect()
print("--- ACTIVITY REPORT ---")
for idx, act in enumerate(report, 1):
    print(f"Log #{idx:02d}: {act}")
```

```python
unique_activities = rdd.distinct().collect()
print("--- UNIQUE ACTIVITY REPORT ---")
for act in unique_activities:
    print(f" - {act}")
```

```python
freq_report = freq_rdd.sortBy(lambda x: x[1], ascending=False).collect()
print("--- FREQUENCY REPORT ---")
for act, count in freq_report:
    print(f"{act:<20} : {count} occurrences")
```

```python
summary_data = freq_rdd.collect()
print("summary")
print(f"Total Platform Actions Recorded : {total_activities}")
print(f"Unique Types of Actions         : {len(summary_data)}")
print(f"{'Activity Type':<20} | {'Count':<5} | {'Percentage':<10}")
for act, count in sorted(summary_data, key=lambda x: x[1], reverse=True):
    pct = (count / total_activities) * 100
    print(f"{act:<20} | {count:<5} | {pct:.2f}%")
```
