```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName('rdd').getOrCreate()
```

```python
numbers=[10,20,30,40,50]
rdd=spark.sparkContext.parallelize(numbers)
rdd.collect()
```

```python
rdd.count()
```

```python
rdd.take(3)
```

```python
result=rdd.map(lambda x:x*x)
result.collect()
```

```python
n=[1,2,3,4,5]
result1=rdd.map(lambda x:x+10)
result1.collect()
```

```python
numbers=[1,2,3,4,5,6,7,8]
rdd=spark.sparkContext.parallelize(numbers)
result=rdd.filter(lambda x:x%2==0)
result.collect()
```

```python
numbers=[10,20,20,30,40,50,10]
rdd=spark.sparkContext.parallelize(numbers)
result=rdd.distinct()
result.collect()
```

```python
rdd1=spark.sparkContext.parallelize([10,20,30,40,50])
rdd2=spark.sparkContext.parallelize([60,70,80,90,100])
result1=rdd1.union(rdd2)
result1.collect()
```

```python
numbers=[40,10,20,50,30]
rdd=spark.sparkContext.parallelize(numbers)
result=rdd.sortBy(lambda x:x)
result.collect()
```

```python
numbers=[10,20,30,40,50]
rdd=spark.sparkContext.parallelize(numbers)
result=rdd.reduce(lambda x,y:x+y)
print(result)
```
