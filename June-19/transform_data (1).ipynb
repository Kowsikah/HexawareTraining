```python
# Read from persistent layers
booking = spark.read.table("airline_operations.airline.bookings_clean")
flight = spark.read.table("airline_operations.airline.flights_clean")

# Run your existing transformations
from pyspark.sql.functions import when, col
booking=booking.withColumnRenamed("ticket_price", "revenue")
booking = booking.withColumn("price_band", when(col("revenue") > 20000, "Premium").when(col("revenue") > 10000, "Standard").otherwise("Budget"))
flight = flight.withColumn("delay_flag", when(col("status") == "Delayed", 1).otherwise(0))

# Overwrite tables with transformed data
booking.write.format("delta").mode("overwrite").saveAsTable("airline_operations.airline.bookings_transformed")
flight.write.format("delta").mode("overwrite").saveAsTable("airline_operations.airline.flights_transformed")
```
