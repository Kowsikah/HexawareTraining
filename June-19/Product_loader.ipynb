```python
product_data=[(101,"Laptop",10000),(102,"Tablet",5000),(103,"Phone",3000),(104,"Watch",1000)]
product_df = spark.createDataFrame(product_data,["id","name","price"])
product_df.show()
```

```python
product_df.createOrReplaceTempView("product")
```

```python
result1=spark.sql("select avg(price) from product")
result1.show()
```
