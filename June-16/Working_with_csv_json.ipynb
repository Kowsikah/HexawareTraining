```python
from pyspark.sql.functions import sum as _sum, coalesce, lit
df_revenue_agg = visits.groupBy("doctor_id").agg(_sum("bill_amount").alias("raw_revenue"))
doctors = doctors.join(df_revenue_agg, on="doctor_id", how="left") \
    .withColumn("revenue", coalesce("raw_revenue", lit(0))) \
    .select("doctor_id", "doctor_name", "specialization", "city", "revenue")

doctors.show()
```

```python
from pyspark.sql.functions import rank
from pyspark.sql.window import Window
window_spec = Window.orderBy(col("revenue").desc())
ranks=doctors.withColumn("rank", rank().over(window_spec))

ranks.show()
```

```python
from pyspark.sql.functions import dense_rank
ranks=doctors.withColumn("rank", dense_rank().over(window_spec))
ranks.show()
```

```python
doctors.show(1)
```

```python
windows_spec=Window.partitionBy("specialization").orderBy(col("revenue"))
ranks=doctors.withColumn("rank", rank().over(windows_spec))
ranks.filter(col("rank")==1).show()
```

```python
windows_spec=Window.partitionBy("specialization").orderBy(col("revenue"))
ranks=doctors.withColumn("rank", rank().over(windows_spec))
ranks.filter(col("rank")<=2).show()
```

```python
window_spec=Window.orderBy(col("revenue").desc()).rowsBetween(Window.unboundedPreceding, Window.currentRow)
running=doctors.withColumn("cum_revenue", sum(col("revenue")).over(window_spec))
running.show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col, lag, lead
window_spec=Window.orderBy(col("revenue").desc())

lag_d=doctors.withColumn("lag", lag(col("revenue")).over(window_spec))
lag_d.show()
from pyspark.sql.functions import lead
lead_d=doctors.withColumn("lead", lead(col("revenue")).over(window_spec))
lead_d.show()
```

```python
city_window_desc = Window.partitionBy("city").orderBy(col("revenue").desc())
city_window_asc = Window.partitionBy("city").orderBy(col("revenue").asc())

print("Highest Earning Doctors in each city:")
doctors.withColumn("rank", rank().over(city_window_desc)).filter(col("rank") == 1).show()

print("Lowest Earning Doctors in each city:")
doctors.withColumn("rank", rank().over(city_window_asc)).filter(col("rank") == 1).show()
```

```python
from pyspark.sql.functions import dense_rank
leaderboard_window = Window.orderBy(col("revenue").desc())
doctors.withColumn("leaderboard_rank", dense_rank().over(leaderboard_window)).show()
```

```python
from pyspark.sql import functions as F
joined_df = doctors.alias("d").join(
    visits.alias("v"),
    F.col("d.doctor_id") == F.col("v.doctor_id"),
    how="inner"
).select(
    "d.doctor_id", "d.doctor_name", "d.specialization", 
    F.col("d.city").alias("doctor_city"),
    "v.visit_id", "v.patient_name", "v.visit_date", "v.diagnosis", "v.bill_amount", "v.payment_"
)
```

```python
doctor_visits_df=joined_df\
    .withColumn("tax", F.col("bill_amount") * 0.05) \
    .withColumn("final_amount", F.col("bill_amount") + F.col("tax")) \
    .withColumn("total_visit_revenue", F.col("bill_amount"))
doctor_visits_df.show()
```

```python
doctor_metrics_df = doctor_visits_df.groupBy("doctor_id", "doctor_name", "specialization") \
    .agg(
        F.count("visit_id").alias("total_patients_treated"),
        F.sum("total_visit_revenue").alias("total_revenue_generated")
    )



```

```python
window_spec_rank = Window.orderBy(F.col("total_revenue_generated").desc())
gold_doctor_rankings_df = doctor_metrics_df.withColumn(
    "doctor_rank", 
    F.dense_rank().over(window_spec_rank)
)
```

```python
gold_specialization_summary_df = doctor_visits_df.groupBy("specialization") \
    .agg(
        F.countDistinct("doctor_id").alias("total_active_doctors"),
        F.count("visit_id").alias("total_visits"),
        F.sum("bill_amount").alias("total_billing_amount"),
        F.sum("total_visit_revenue").alias("total_departmental_revenue")
    )
```

```python
doctor_visits_df.write.mode("overwrite").parquet(
    "/Volumes/practice/practice/filestorage/silver_doctor_visits.parquet"
)
```

```python
gold_doctor_rankings_df.write.mode("overwrite") \
    .saveAsTable("practice.practice.gold_doctor_performance_report")

gold_specialization_summary_df.write.mode("overwrite") \
    .saveAsTable("practice.practice.gold_specialization_summary_report")
```

```python
clean_hospital_df = hospital_config.withColumn(
    "join_spec", F.lower(F.trim(F.col("service")))
)
clean_summary_df = gold_specialization_summary_df.withColumn(
    "join_spec", F.lower(F.trim(F.col("specialization")))
)

dashboard_dataset_df = clean_hospital_df.join(
    clean_summary_df, 
    on="join_spec", 
    how="left"
).select(
    F.col("hospital_id"),
    F.col("hospital_name"),
    F.col("city"),
    F.col("service"),
    F.coalesce(F.col("total_active_doctors"), F.lit(0)).alias("total_allocated_doctors"),
    F.coalesce(F.col("total_visits"), F.lit(0)).alias("total_completed_visits"),
    F.coalesce(F.col("total_departmental_revenue"), F.lit(0.0)).alias("total_revenue")
)

# Save the final dashboard view as an accessible production table
dashboard_dataset_df.write.mode("overwrite") \
    .saveAsTable("practice.practice.dashboard_hospital_operations")

# Display a preview of the dataset
dashboard_dataset_df.show(truncate=False)
```
