```python
from pyspark.sql import SparkSession
spark=SparkSession.builder\
      .appName("PySpark Day---1")\
      .getOrCreate()
```

```python
print(spark)
```

```python
data = [
    (101, "Rahul Sharma", "Data Engineering", 75000, "Hyderabad"),
    (102, "Priya Reddy", "AI Engineering", 85000, "Bangalore"),
    (103, "Amit Kumar", "Data Engineering", 65000, "Mumbai"),
    (104, "Sneha Patel", "Data Science", 95000, "Chennai"),
    (105, "Farhan Ali", "Cloud Engineering", 80000, "Delhi")
]
```

```python
columns = [
    "employee_id",
    "name",
    "department",
    "salary",
    "city"
]
```

```python
df=spark.createDataFrame(data,columns)
df.show()
```

```python
df.select("name","salary").show()
```

```python
df.printSchema()
```

```python
df.filter(df.salary>80000).show()
```

```python
df.filter(df.city=="Hyderabad").show()
```

```python
df.orderBy("salary").show()
```

```python
df.orderBy(df.salary.desc()).show()
```

```python
from pyspark.sql.functions import col
df_bonus=df.withColumn("bonus",col("salary")*0.10)
df_bonus.show()
```

```python
df_renamed=df.withColumnRenamed("salary","monthly_salary")
df_renamed.show()
```

```python
df.groupBy("department").count().show()
```

```python
new_df=df.withColumn("Bonus",df.salary*0.10)
new_df.show()
```

```python
updated_df=new_df.withColumn("tax",df.salary*0.05)
updated_df.show()
```

```python
updated_df=updated_df.withColumn("Annual_Salary",df.salary*12)
updated_df.show()
```

```python
updated_df=updated_df.withColumn("salary_after_bonus",updated_df.salary+updated_df.Bonus)
updated_df.show()
```

```python
from pyspark.sql.functions import when
updated_df=updated_df.withColumn("salary_category",when(updated_df.monthly_salary>=85000,"High").when(updated_df.monthly_salary>=75000,"Medium").otherwise("Low"))
updated_df.show()
```
