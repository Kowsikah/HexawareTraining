```python
df=df.withColumn("tax",df.ticket_price*0.05)
df.show()
```

```python
df=df.withColumn("Final_Price",df.ticket_price+df.tax)
df.show()
```

```python
from pyspark.sql.functions import when
df=df.withColumn("Category",when((df.ticket_price)>30000,"Premium").when(df.ticket_price>=10000,"Standard").otherwise("Budget"))
df.show()
```

```python
df.groupBy("Category").count().show()
```

```python
confirmed_df=df.filter(df.status=="Confirmed")
confirmed_df.write.parquet("confirmed_flights2.parquet")
```

```python
hotels_df = spark.read.option(
"multiline",
"true"
).json("hotels.json")
hotels_df.show(truncate=False)
hotels_df.printSchema()
```

```python
hotels_df=hotels_df.withColumn("total_potential_revenue",hotels_df.rooms_available*hotels_df.price_per_night)
hotels_df.show()
```

```python
hotels_df.orderBy(hotels_df.rating.desc()).show()
```

```python
from pyspark.sql.functions import col, concat_ws

hotels_flattened_df = hotels_df.select(
    col("hotel_id"),
    col("hotel_name"),
    col("city"),
    col("category"),
    col("rating"),
    col("rooms_available"),
    col("price_per_night"),
    concat_ws(", ", col("amenities")).alias("amenities_list"), # Flatten amenities array to string
    col("contact.phone").alias("contact_phone"), # Flatten nested phone
    col("contact.email").alias("contact_email"),  # Flatten nested email
    col("total_potential_revenue")
)

hotels_flattened_df.show(truncate=False)
hotels_flattened_df.printSchema()
```

```python
hotels_flattened_df.write.mode("overwrite").parquet("hotels_flattened.parquet")
```
