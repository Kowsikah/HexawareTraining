```python
from pyspark.sql.window import Window
from pyspark.sql.functions import rank, desc, col
temp_p = payments.select(payments.customer_id.alias("p_customer_id"), payments.amount_paid.alias("p_amount_paid"))
temp_c = customer.select(customer.customer_id.alias("c_customer_id"), customer.customer_name)
joined_customer_payment_resolved = temp_p.join(temp_c, temp_p.p_customer_id == temp_c.c_customer_id, "inner") \
                                     .select(col("c_customer_id").alias("cust_id"),
                                             col("customer_name"),
                                             col("p_amount_paid").alias("amount_paid"))
window_spec = Window.orderBy(desc(col("amount_paid")))
customer_amount_rank = joined_customer_payment_resolved.withColumn("customer_rank", rank().over(window_spec))
customer_amount_rank.select(
    col("cust_id").alias("customer_id"),
    col("customer_name"),
    col("amount_paid"),
    col("customer_rank")
).orderBy(col("customer_rank")).show()
```

```python
from pyspark.sql.functions import sum, month, col
from pyspark.sql.window import Window
payments_with_customer_id = payments.select(payments.customer_id, payments.bill_month, payments.amount_paid)
monthly_revenue = payments_with_customer_id.groupBy(month("bill_month").alias("month")) \
                                        .agg(sum("amount_paid").alias("monthly_total_revenue")) \
                                        .orderBy("month")
window_spec_running_total = Window.orderBy("month").rowsBetween(Window.unboundedPreceding, Window.currentRow)
running_total_revenue = monthly_revenue.withColumn(
    "running_total_revenue",
    sum("monthly_total_revenue").over(window_spec_running_total)
)

running_total_revenue.show()
```

```python
from pyspark.sql.functions import lag, lead, col, month
from pyspark.sql.window import Window
usage_for_lag_lead = usage.select(col("customer_id"), col("usage_month"), col("data_used_gb")) \
                            .orderBy(col("customer_id"), col("usage_month"))
window_spec_lag_lead = Window.partitionBy("customer_id").orderBy("usage_month")
lag_lead_usage = usage_for_lag_lead.withColumn(
    "previous_month_data_used_gb",
    lag(col("data_used_gb"), 1).over(window_spec_lag_lead)
).withColumn(
    "next_month_data_used_gb",
    lead(col("data_used_gb"), 1).over(window_spec_lag_lead)
)

lag_lead_usage.show()
```

```python
from pyspark.sql.functions import col
customers_with_increased_usage = lag_lead_usage.filter(
    (col("data_used_gb") > col("previous_month_data_used_gb")) & col("previous_month_data_used_gb").isNotNull()
)

customers_with_increased_usage.select("customer_id", month("usage_month").alias("month"), "previous_month_data_used_gb", "data_used_gb").show()
```

```python
import pandas as pd
incremental_usage_data = [
    (1016, 101, '2026-03-01', 60, 1100, 140),
    (1017, 102, '2026-03-01', 38, 700, 90),
    (1018, 104, '2026-03-01', 65, 1300, 170),
    (1019, 105, '2026-03-01', 80, 1600, 220)
]

incremental_usage_schema = [
    "usage_id", "customer_id", "usage_month", "data_used_gb", "call_minutes", "sms_count"
]

pd_incremental_usage_df = pd.DataFrame(incremental_usage_data, columns=incremental_usage_schema)
incremental_usage_spark_df = spark.createDataFrame(pd_incremental_usage_df)
from pyspark.sql.functions import to_timestamp
incremental_usage_spark_df = incremental_usage_spark_df.withColumn("usage_month", to_timestamp(col("usage_month")))

print("Incremental usage data created:")
incremental_usage_spark_df.show()
```

```python
from pyspark.sql.functions import col
from pyspark.sql.types import IntegerType
incremental_usage_spark_df = incremental_usage_spark_df.withColumn("usage_id", col("usage_id").cast(IntegerType()))
incremental_usage_spark_df = incremental_usage_spark_df.withColumn("customer_id", col("customer_id").cast(IntegerType()))
incremental_usage_spark_df = incremental_usage_spark_df.withColumn("data_used_gb", col("data_used_gb").cast(IntegerType()))
incremental_usage_spark_df = incremental_usage_spark_df.withColumn("call_minutes", col("call_minutes").cast(IntegerType()))
incremental_usage_spark_df = incremental_usage_spark_df.withColumn("sms_count", col("sms_count").cast(IntegerType()))

print("Casted incremental usage data to match existing usage DataFrame schema:")
incremental_usage_spark_df.printSchema()
```

```python

inc_u_alias = incremental_usage_spark_df.alias("inc_u")
c_alias = customer.alias("c")
p_alias = plan.alias("p")
incremental_with_plan_customer = inc_u_alias.join(
    c_alias, col("inc_u.customer_id") == col("c.customer_id"), "left"
).join(
    p_alias, col("c.plan_id") == col("p.plan_id"), "left"
)

processed_incremental_usage = incremental_with_plan_customer.withColumn(
    "temp_data_quality_status",
    when(col("inc_u.data_used_gb").isNull(), "Incomplete").otherwise("Complete")
).withColumn(
    "temp_usage_category",
    when(col("inc_u.data_used_gb") > 70, "Heavy User")
    .when(col("inc_u.data_used_gb") >= 30, "Medium User")
    .otherwise("Low User")
).withColumn(
    "temp_calculated_over_usage_gb", col("inc_u.data_used_gb") - col("p.data_limit_gb")
).withColumn(
    "temp_over_usage_gb",
    when(col("temp_calculated_over_usage_gb") > 0, "Yes").otherwise("No")
).select(
    col("inc_u.usage_id"),
    col("inc_u.customer_id"),
    col("inc_u.usage_month"),
    col("inc_u.data_used_gb"),
    col("inc_u.call_minutes"),
    col("inc_u.sms_count"),
    col("temp_usage_category").alias("usage_category"),
    col("temp_over_usage_gb").alias("over_usage_gb"),
    col("temp_data_quality_status").alias("data_quality_status")
)

usage = usage.unionByName(processed_incremental_usage)

print("Incremental records added to the 'usage' DataFrame.")
usage.show()
```

```python
from pyspark.sql.functions import when, col


customer_temp_alias_1 = customer.alias("c_tmp1").withColumnRenamed("plan_id", "customer_plan_id_tmp")
plan_temp_alias_1 = plan.alias("pl_tmp1")


usage_for_join_alias = usage.alias("u_join").withColumnRenamed("customer_id", "usage_customer_id_unique")


updated_usage_temp_df = usage_for_join_alias.join(
    customer_temp_alias_1, usage_for_join_alias.usage_customer_id_unique == customer_temp_alias_1.customer_id, "inner"
).join(
    plan_temp_alias_1, customer_temp_alias_1.customer_plan_id_tmp == plan_temp_alias_1.plan_id, "left"
)


updated_usage_with_over_usage = updated_usage_temp_df.withColumn(
    "calculated_over_usage_gb", col("data_used_gb") - col("data_limit_gb")
).withColumn(
    "over_usage_gb_indicator", when(col("calculated_over_usage_gb") > 0, "Yes").otherwise("No")
).withColumn(
    "data_quality_status_calc", when(col("data_used_gb").isNull(), "Incomplete").otherwise("Complete")
)


usage = updated_usage_with_over_usage.select(
    updated_usage_with_over_usage["usage_id"],
    # Select the uniquely named customer_id from the usage side of the join
    updated_usage_with_over_usage["usage_customer_id_unique"].alias("customer_id"),
    updated_usage_with_over_usage["usage_month"],
    updated_usage_with_over_usage["data_used_gb"],
    updated_usage_with_over_usage["call_minutes"],
    updated_usage_with_over_usage["sms_count"],
    updated_usage_with_over_usage["usage_category"],
    col("over_usage_gb_indicator").alias("over_usage_gb"),
    col("data_quality_status_calc").alias("data_quality_status")
)


customer_gold_alias = customer.alias("c_gold").withColumnRenamed("plan_id", "customer_plan_id")
usage_gold_alias = usage.alias("u_gold")
payments_gold_alias = payments.alias("p_gold")

clean_plan_for_gold_recreation = spark.read.parquet("Silver/plan")
plan_gold_alias = clean_plan_for_gold_recreation.alias("pl_gold")

joined_gold_df_after_incremental = customer_gold_alias.join(usage_gold_alias, customer_gold_alias.customer_id == usage_gold_alias.customer_id, "inner") \
                                           .join(payments_gold_alias, (customer_gold_alias.customer_id == payments_gold_alias.customer_id) & (usage_gold_alias.usage_month == payments_gold_alias.bill_month), "inner") \
                                           .join(plan_gold_alias, customer_gold_alias.customer_plan_id == plan_gold_alias.plan_id, "inner")

recreated_gold_output = joined_gold_df_after_incremental.select(
    customer_gold_alias.customer_id,
    customer_gold_alias.customer_name,
    customer_gold_alias.city,
    customer_gold_alias.state,
    customer_gold_alias.age,
    customer_gold_alias.gender,
    customer_gold_alias.status,
    customer_gold_alias.data_quality_status.alias("customer_data_quality_status"),
    usage_gold_alias.usage_id,
    usage_gold_alias.usage_month,
    usage_gold_alias.data_used_gb,
    usage_gold_alias.call_minutes,
    usage_gold_alias.sms_count,
    usage_gold_alias.usage_category,
    usage_gold_alias.over_usage_gb,
    usage_gold_alias.data_quality_status.alias("usage_data_quality_status"),
    payments_gold_alias.payment_id,
    payments_gold_alias.bill_month,
    payments_gold_alias.amount_paid,
    payments_gold_alias.payment_mode,
    payments_gold_alias.payment_status,
    payments_gold_alias.payment_catgeory,
    payments_gold_alias.churn_risk,
    payments_gold_alias.data_quality_status.alias("payment_data_quality_status"),
    plan_gold_alias.plan_id.alias("actual_plan_id"),
    plan_gold_alias.plan_name,
    plan_gold_alias.monthly_fee,
    plan_gold_alias.data_limit_gb,
    plan_gold_alias.features.ott_included,
    plan_gold_alias.features.unlimited_calls,
    plan_gold_alias.features.roaming
)

recreated_gold_output.write.mode("overwrite").partitionBy("usage_month").parquet("Gold/customer_telecom_analytics")

print("Gold output recreated successfully with incremental data.")
```

```python
from pyspark.sql.functions import col, count, sum, avg
c_alias = customer.alias("c")
u_alias = usage.alias("u")
p_alias = payments.alias("p")
plan_fresh = spark.read.parquet("plan")
pl_alias = plan_fresh.alias("pl")
joined_data_for_plan_report = c_alias.join(u_alias, c_alias.customer_id == u_alias.customer_id, "inner") \
                                     .join(p_alias, c_alias.customer_id == p_alias.customer_id, "inner") \
                                     .join(pl_alias, c_alias.plan_id == pl_alias.plan_id, "inner") \
                                     .select(
                                         pl_alias.plan_name,
                                         c_alias.customer_id,
                                         u_alias.data_used_gb,
                                         p_alias.amount_paid
                                     )

plan_performance_report = joined_data_for_plan_report.groupBy(col("plan_name")) \
                                                  .agg(
                                                      count(col("customer_id")).alias("total_customers"),
                                                      sum(col("data_used_gb")).alias("total_data_usage"),
                                                      avg(col("data_used_gb")).alias("average_data_usage"),
                                                      sum(col("amount_paid")).alias("total_revenue")
                                                  )

plan_performance_report.orderBy(col("total_revenue").desc()).show()
```
